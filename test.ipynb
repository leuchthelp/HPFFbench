```python
import xarray as xr

# Sample data
customer1 = xr.DataArray(data=["Jane Smith"], dims=["name"])
customer2 = xr.DataArray(data=["John Doe"], dims=["name"])


trans1 = xr.DataArray(data=[["1", "2"], [100, 150]], dims=["Date", "Amount"])

trans2 = xr.DataArray(data=[["1"], [200]], dims=["Date", "Amount"])



ds1 = xr.Dataset({"customers": customer1, "transactions": trans1})
ds2 = xr.Dataset({"customers": customer2, "transactions": trans2})

dt = xr.DataTree.from_dict({"child1": ds1, "child2": ds2})

test = dt.to_zarr("temp.zarr")

dt.compute()
```
